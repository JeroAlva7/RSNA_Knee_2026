```python
from pathlib import Path

INPUT_DIR = Path("/kaggle/input")

print("Contenido de /kaggle/input:")

for item in INPUT_DIR.iterdir():
    print(item)

print("\nBuscando test.csv...")

test_files = list(INPUT_DIR.rglob("test.csv"))

for file in test_files:
    print(file)

print("\nCantidad de test.csv encontrados:", len(test_files))
```

```python
import pandas as pd

test_path = test_files[0]

print("Ruta utilizada:")
print(test_path)

test = pd.read_csv(test_path)

print("\nForma del test:")
print(test.shape)

print("\nContenido:")
print(test)
```

```python
TARGETS = [
    "ACL",
    "MCL",
    "Medial Meniscus",
    "Lateral Meniscus",
    "Medial OA",
    "Lateral OA",
    "PF OA",
    "Effusion",
    "Synovitis",
    "Baker's",
    "Contusion",
    "Fracture",
]

submission = test[["StudyInstanceUID"]].copy()

for target in TARGETS:
    submission[target] = 0.5

print(submission)
print("\nForma:", submission.shape)
```

```python
expected_columns = ["StudyInstanceUID"] + TARGETS

assert list(submission.columns) == expected_columns
assert submission["StudyInstanceUID"].is_unique
assert submission[TARGETS].notna().all().all()

assert (
    (submission[TARGETS] >= 0)
    & (submission[TARGETS] <= 1)
).all().all()

print("Submission válida.")
```

```python
OUTPUT_PATH = "/kaggle/working/submission.csv"

submission.to_csv(
    OUTPUT_PATH,
    index=False
)

print("Archivo creado:")
print(OUTPUT_PATH)

print("\nVista previa:")
print(pd.read_csv(OUTPUT_PATH).head())
```
